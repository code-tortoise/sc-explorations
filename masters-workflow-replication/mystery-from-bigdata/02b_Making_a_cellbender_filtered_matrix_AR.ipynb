# Read in cellbender lane outs for F137

```python
adata_list1 = []
for i in list(range(36,81)):
    n = 'WS_wEMB102023'+str(i)
    
    if (n == 'WS_wEMB10202360'): # Lanes not passed cellranger therefore no cellbender output
        continue
    
    print(f'Loading in data: {n}')
    data = sc.read_10x_h5(cellbender_outputs + n + '/outs' + '/cellbender_out_filtered.h5')
    print('Making variable names unique')
    data.var_names_make_unique()
    data.obs['sequencing_lane_ID'] = n
    adata_list1.append(data)
    print(f'Data {n} loaded successfully. Data shape is {data.shape}')
    print("")
    del data # to try and prevent stackoverflow

Embryo1 = sc.AnnData.concatenate(*adata_list1, join = 'inner', batch_key = None, batch_categories = None, index_unique = None)
Embryo1.obs.index = Embryo1.obs.index.str.replace('-1','',regex=True)
Embryo1.obs.index = Embryo1.obs.index + '-' + Embryo1.obs['sequencing_lane_ID']
#Embryo1.var['genome'] = 'GRCh38'
print("")
print(f'Embryo1 made successfully')
print("")
```

# Calculate for F137

```python
Embryo1_cellbender_matricies = []

Embryo1_outs_comp = pd.DataFrame(columns=('Sample', 'n_cells', 'cellranger reads raw', 'cellbender reads filt', 'percent filtered'))
for i in list(range(36,81)):
    n = 'WS_wEMB102023'+str(i)
    
    if (n == 'WS_wEMB10202360'): # Lanes not passed cellranger therefore no cellbender output
        continue
    
    print(f'Starting loop for sample: {n}')
    print('Reading in cellranger raw matrix:')
    cellranger_raw_ = sc.read_10x_h5('/home/jovyan/mounting_point/Newcastle_1TB/newcastle/AntonyRose/uncompressed_embryo/embryo1/' + n + '/raw_feature_bc_matrix.h5')
    print('Making variable names unique')
    cellranger_raw_.var_names_make_unique()
    print('Cell ranger raw matrix loaded successfully')
    print()
    print('Reading in cell bender filtered matrix:')
    cellbender_filt = sc.read_10x_h5('/home/jovyan/mounting_point/Newcastle_1TB/newcastle/AntonyRose/tic-1254-results/' + n + '/outs/cellbender_out_filtered.h5')
    print('Making variable names unique')
    cellbender_filt.var_names_make_unique()
    print('Cell bender filt matrix loaded successfully')
    print()
    print('Proceeding to compare the two matricies counts values:')
    
    cellranger_raw_ = cellranger_raw_[cellranger_raw_.obs.index.isin(list(cellbender_filt.obs.index))]
    n_cells = len(cellbender_filt.obs)
    
    data = cellbender_filt
    UMI_filt = pd.DataFrame(data.X.todense())
    #UMI_filt.index = data.obs.index
    #UMI_filt.columns = data.var.index
    UMI_filt_sum = sum(UMI_filt.sum(axis=1))
    
    data = cellranger_raw_
    UMI_raw = pd.DataFrame(data.X.todense())
    #UMI_raw.index = data.obs.index
    #UMI_raw.columns = data.var.index
    UMI_raw_sum = sum(UMI_raw.sum(axis=1))
    
    if (UMI_filt_sum == UMI_raw_sum):
        raise Exception('Both matricies match - not good!!!')
    
    percent_filt = round(((UMI_raw_sum-UMI_filt_sum)/UMI_raw_sum)*100,2)
    print(f'Stats for sample {n}: {n_cells} cells, {UMI_raw_sum} reads (raw), {UMI_filt_sum} reads (filt), {percent_filt}% filtered')
    table = pd.DataFrame({'Sample':[n], 'n_cells':[n_cells], 'cellranger reads raw':[UMI_raw_sum], 'cellbender reads filt':[UMI_filt_sum], 'percent filtered':[percent_filt]})
    Embryo1_outs_comp = pd.concat([Embryo1_outs_comp,table])
    print()
    print('Loop complete')
    print()
    #del cellranger_raw_
    #del cellbender_filt
    #del data
    #del UMI_raw
    #del UMI_filt
```

# summarise per section

```python
data = Embryo3_outs_comp[:]
data_name = 'Embryo3_outs_comp'
col = 'Section'

for i in data[col].unique():
    print(f'{data_name} for {i}')
    print()
    df = data[data[col].isin([i])]
    n_cells = df['n_cells'].sum()
    reads_raw = df['cellranger reads raw'].sum()
    reads_filt = df['cellbender reads filt'].sum()
    percent = df['percent filtered'].mean()
    section = i
    section_indiv = 'all'
    no_reads_removed = df['no_reads_removed'].sum()
    sum_row = {'Sample':'Combined', 'n_cells':n_cells,'cellranger reads raw':reads_raw,'cellbender reads filt':reads_filt,'percent filtered':percent,'Section':section,'Section_indiv':section_indiv,'no_reads_removed':no_reads_removed}
    df = df.append(sum_row, ignore_index=True)
    print(df)
    print()
    print()
    print()
```
